### Introduction

This notebook lets you map **wheat, rapeseed and maize** for any area in a country that has a CropLogic model.

- The models are **season-specific, in-season and interpretable**.
- The mapping uses **selected bands and vegetation indices** from the literature that are known to separate the crop well (for example, a spike in NDYI during rapeseed flowering).
- The algorithm runs in a **sliding window**, so you get an in-season map based on the dates you specify.
- The crop-type layer has a **third band** which contains the median dekad from the season start at which the phenological event was confirmed and the model assigned the pixel to the crop.
- The app picks the **12 months** that match the season and the crop, starts the correct model on the **openEO cloud**, and shows the result on a map when the job is finished.

### 1. Create your crop type map

**Step 1: Select a country**

Click a blue country on the map or choose it in the list. The map zooms to the country.

**Step 2: Choose the season and the crops**

Each country has one or two growing seasons. Pick the season, then tick the crops you want to map.<br>
The notes under the crops tell you how much to trust each model: its expected F1 score, and whether it was trained on this country or borrowed from a similar one.

**Step 3: Select your area**

Zoom in until the **10 x 10 km grid** appears and click a cell, or draw a rectangle or polygon yourself.<br>
The size of the area is shown below the map.

<div class="alert alert-block alert-info">
<b>Area limit:</b><br>
Areas up to 100 km² run as one job. A larger area is split into tiles of at most 100 km², each tile is run as its own job, and the tiles are shown together on one map. More tiles use more credits and time, so we recommend you first run the model on a small area to check the result before you map more.
</div>

Optionally, tick **Show the WorldCereal crop calendar points** to compare the season of the model with the WorldCereal crop calendar at the same place.

**Step 4: Choose the year and run**

Choose the year in which the season ends (the harvest year) and press **Run**.

<div class="alert alert-block alert-warning">
<b>Processing time and credits:</b><br>
A job takes about 10 minutes and uses your free openEO credits. If the season is not finished yet, the map uses the data available so far.<br>
To stop, interrupt the kernel and cancel the job in the <a href='https://openeo.dataspace.copernicus.eu/' target='_blank' rel='noopener'>CDSE dashboard</a>.
</div>

**Step 5: Inspect your map**

When the job is finished, the result is drawn on the map. Use the buttons under the map to switch between the **crop map**, the **probability** and the **in-season date** (when in the season the crop was first recognised).<br>
The files are saved in the `results` folder next to this notebook.
